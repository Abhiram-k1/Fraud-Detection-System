```python
# ============================================================
# Credit Card Fraud Detection — Academic ML Pipeline
# ✅ Google Colab Compatible Version
# Integrates: Toeplitz Feature Engineering, Gradient Descent,
#             SGD-trained Neural Network, XGBoost Comparison
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt          # ← NO 'Agg' backend; Colab shows inline
import seaborn as sns
import warnings, os
warnings.filterwarnings('ignore')

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (classification_report, roc_auc_score,
                              roc_curve, auc, accuracy_score,
                              precision_score, recall_score, f1_score)
from sklearn.utils import resample
from xgboost import XGBClassifier
from IPython.display import display       # for pretty DataFrame tables in Colab

# ── Output folder (local Colab runtime) ───────────────────────
OUTPUT_DIR = '/content/fraud_outputs'
os.makedirs(OUTPUT_DIR, exist_ok=True)

def _save_and_show(filename: str):
    """Save current figure to OUTPUT_DIR and display inline."""
    path = os.path.join(OUTPUT_DIR, filename)
    plt.savefig(path, dpi=150, bbox_inches='tight')
    plt.show()                            # ← this line makes it appear in Colab
    plt.close()
    print(f"   💾 Saved → {path}\n")


# ============================================================
# SECTION 1: DATA LOADING
# ============================================================

def load_data(filepath: str = '/content/creditcard.csv') -> pd.DataFrame:
    df = pd.read_csv(filepath)
    print(f"[load_data] Raw shape      : {df.shape}")
    print(f"[load_data] Missing values : {df.isna().sum().sum()}")
    print(f"[load_data] Class counts   :\n{df['Class'].value_counts()}\n")
    df = df.dropna()
    print(f"[load_data] Shape after dropna: {df.shape}\n")
    return df


# ============================================================
# SECTION 2: EDA — visualised inline + saved
# ============================================================

def eda(df: pd.DataFrame) -> list:
    threshold = 0.13

    # ── Plot 1: Class imbalance ────────────────────────────────
    fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    fig.suptitle("Exploratory Data Analysis", fontsize=15, fontweight='bold')

    counts = df['Class'].value_counts()
    axes[0].bar(['Legit (0)', 'Fraud (1)'], counts.values,
                color=['steelblue', 'crimson'], edgecolor='black')
    axes[0].set_title("Class Distribution (Imbalanced)", fontsize=13)
    axes[0].set_ylabel("Count")
    for i, v in enumerate(counts.values):
        axes[0].text(i, v * 1.01, f'{v:,}', ha='center', fontsize=11)

    # ── Plot 2: Feature correlation with target ────────────────
    corr = df.corr()['Class'].drop('Class')
    bar_colors = ['crimson' if abs(c) >= threshold else 'steelblue'
                  for c in corr.values]
    axes[1].bar(corr.index, corr.values, color=bar_colors, edgecolor='black')
    axes[1].axhline( threshold, color='green', linestyle='--',
                     linewidth=1.5, label=f'±{threshold} threshold')
    axes[1].axhline(-threshold, color='green', linestyle='--', linewidth=1.5)
    axes[1].set_title("Feature Correlation with Class", fontsize=13)
    axes[1].set_xticklabels(corr.index, rotation=90, fontsize=8)
    axes[1].legend(fontsize=9)

    plt.tight_layout()
    _save_and_show('1_eda_plots.png')

    # ── Plot 3: Correlation heatmap of selected features ───────
    selected = [c for c in corr.index if abs(corr[c]) >= threshold]
    df_sub = df[selected + ['Class']]
    plt.figure(figsize=(10, 7))
    sns.heatmap(df_sub.corr(), annot=True, fmt='.2f',
                cmap='coolwarm', linewidths=0.5, annot_kws={'size': 8})
    plt.title("Correlation Heatmap — Selected Features", fontsize=13)
    plt.tight_layout()
    _save_and_show('2_correlation_heatmap.png')

    print(f"[eda] Selected {len(selected)} features: {selected}\n")
    return selected


# ============================================================
# SECTION 3: TOEPLITZ-BASED FEATURE ENGINEERING  (UNIT 1)
# ============================================================

def create_toeplitz_features(df: pd.DataFrame,
                              columns: list,
                              lags: int = 3) -> pd.DataFrame:
    """
    ── Why Toeplitz? ──────────────────────────────────────────
    A Toeplitz matrix T has constant diagonals: T[i,j] = t[i-j].
    Applying an FIR filter to signal x means computing y = T·x.
    Each row of T is the SAME kernel shifted by one position —
    this is shift-invariance.

    Building lag columns [x(t-1), x(t-2), x(t-3)] for every
    timestep t constructs exactly those Toeplitz rows.
    Fraud often appears as a sudden spike in Amount vs. recent
    history — a pattern this sliding-window structure captures.
    """
    df_out = df.copy()
    for col in columns:
        for lag in range(1, lags + 1):
            # .shift(lag) selects the (t-lag) entry — one Toeplitz row element
            df_out[f'{col}_lag{lag}'] = df_out[col].shift(lag)
    df_out.dropna(inplace=True)
    print(f"[toeplitz] Added {lags} lags for {columns}. Shape: {df_out.shape}\n")
    return df_out


# ============================================================
# SECTION 4: DATA PREPARATION
# ============================================================

def preprocess_data(df: pd.DataFrame, feature_cols: list, test_size=0.2):
    X = df[feature_cols].values
    y = df['Class'].values

    X_tr, X_te, y_tr, y_te = train_test_split(
        X, y, test_size=test_size, random_state=42, stratify=y)

    # Undersample INSIDE training split only — never touch the test set
    tr = pd.DataFrame(X_tr, columns=feature_cols)
    tr['Class'] = y_tr
    majority     = tr[tr['Class'] == 0]
    minority     = tr[tr['Class'] == 1]
    majority_dn  = resample(majority, replace=False,
                            n_samples=len(minority), random_state=42)
    balanced     = pd.concat([majority_dn, minority]).sample(frac=1, random_state=42)

    X_tr_bal = balanced[feature_cols].values
    y_tr_bal = balanced['Class'].values

    print(f"[preprocess] Train balanced : {X_tr_bal.shape}  "
          f"(Fraud={int(y_tr_bal.sum())})")
    print(f"[preprocess] Test original  : {X_te.shape}  "
          f"(Fraud={int(y_te.sum())})\n")

    scaler      = StandardScaler()
    X_tr_scaled = scaler.fit_transform(X_tr_bal)
    X_te_scaled = scaler.transform(X_te)
    return X_tr_scaled, X_te_scaled, y_tr_bal, y_te, scaler


# ============================================================
# SECTION 5-A: LOGISTIC REGRESSION — GRADIENT DESCENT  (UNIT 2)
# ============================================================

class LogisticRegressionGD:
    """
    Full-batch Gradient Descent on Binary Cross-Entropy loss.

    Loss:     L = -[y·log(ŷ) + (1-y)·log(1-ŷ)]
    Gradient: ∇w = (1/m) Xᵀ(ŷ - y)
    Update:   w ← w - α·∇w

    Full-batch means every sample contributes to ONE gradient
    before a weight update → stable convergence, smooth loss curve.
    """
    def __init__(self, lr=0.1, n_iter=1000, verbose=True):
        self.lr, self.n_iter, self.verbose = lr, n_iter, verbose
        self.weights = self.bias = None
        self.losses = []

    @staticmethod
    def _sigmoid(z):
        return np.where(z >= 0, 1/(1+np.exp(-z)), np.exp(z)/(1+np.exp(z)))

    @staticmethod
    def _bce(y, yh):
        eps = 1e-9
        return -np.mean(y*np.log(yh+eps) + (1-y)*np.log(1-yh+eps))

    def fit(self, X, y):
        m, n = X.shape
        self.weights = np.zeros(n)
        self.bias = 0.0
        for i in range(self.n_iter):
            yh    = self._sigmoid(X @ self.weights + self.bias)
            loss  = self._bce(y, yh)
            self.losses.append(loss)
            err   = yh - y
            self.weights -= self.lr * (X.T @ err) / m
            self.bias    -= self.lr * err.mean()
            if self.verbose and (i+1) % 200 == 0:
                print(f"   [GD iter {i+1:4d}]  BCE = {loss:.4f}")
        return self

    def predict_proba(self, X):
        return self._sigmoid(X @ self.weights + self.bias)

    def predict(self, X, thr=0.5):
        return (self.predict_proba(X) >= thr).astype(int)


def train_logistic_gd(X_train, y_train, lr=0.1, n_iter=800):
    print("[train_logistic_gd] Training Logistic Regression (GD)...")
    m = LogisticRegressionGD(lr=lr, n_iter=n_iter)
    m.fit(X_train, y_train)
    print()
    return m


# ============================================================
# SECTION 5-B: NEURAL NETWORK — SGD  (UNIT 2)
# ============================================================

class NeuralNetworkSGD:
    """
    One hidden layer: Input → ReLU → Sigmoid output.
    Trained with true Stochastic Gradient Descent (one sample/update).

    SGD vs GD:
      GD  → average gradient over ALL samples → one update
      SGD → gradient from ONE sample → immediate update
      SGD is noisier but faster and can escape shallow local minima.

    Backprop chain rule:
      δ_out    = (ŷ - y)                      ← output error
      δ_hidden = (W2ᵀ · δ_out) ⊙ ReLU'(z1)  ← hidden error
    """
    def __init__(self, input_size, hidden_size=16, lr=0.01,
                 epochs=20, verbose=True):
        self.lr, self.epochs, self.verbose = lr, epochs, verbose
        self.losses = []
        s1 = np.sqrt(2.0 / input_size)
        s2 = np.sqrt(2.0 / hidden_size)
        self.W1 = np.random.randn(input_size, hidden_size) * s1
        self.b1 = np.zeros(hidden_size)
        self.W2 = np.random.randn(hidden_size, 1) * s2
        self.b2 = np.zeros(1)

    @staticmethod
    def _relu(z):       return np.maximum(0, z)
    @staticmethod
    def _relu_d(z):     return (z > 0).astype(float)
    @staticmethod
    def _sigmoid(z):
        return np.where(z >= 0, 1/(1+np.exp(-z)), np.exp(z)/(1+np.exp(z)))

    def _forward(self, x):
        z1 = x @ self.W1 + self.b1
        a1 = self._relu(z1)
        z2 = a1 @ self.W2 + self.b2
        return z1, a1, self._sigmoid(z2)

    def fit(self, X, y):
        np.random.seed(42)
        m = X.shape[0]
        for epoch in range(self.epochs):
            idx = np.random.permutation(m)
            X_s, y_s = X[idx], y[idx]
            epoch_loss = 0.0
            for i in range(m):
                xi = X_s[i:i+1]; yi = y_s[i]
                z1, a1, yh = self._forward(xi)
                yhs = yh[0, 0]
                eps = 1e-9
                epoch_loss += -(yi*np.log(yhs+eps) + (1-yi)*np.log(1-yhs+eps))

                d_out  = yhs - yi
                dW2    = a1.T * d_out
                db2    = np.array([d_out])
                d_hid  = (self.W2 * d_out).T * self._relu_d(z1)
                dW1    = xi.T @ d_hid
                db1    = d_hid[0]

                self.W2 -= self.lr * dW2
                self.b2 -= self.lr * db2
                self.W1 -= self.lr * dW1
                self.b1 -= self.lr * db1

            avg = epoch_loss / m
            self.losses.append(avg)
            if self.verbose and (epoch+1) % 5 == 0:
                print(f"   [SGD epoch {epoch+1:3d}/{self.epochs}]  BCE = {avg:.4f}")
        return self

    def predict_proba(self, X):
        _, _, yh = self._forward(X)
        return yh.ravel()

    def predict(self, X, thr=0.5):
        return (self.predict_proba(X) >= thr).astype(int)


def train_neural_network(X_train, y_train, hidden_size=16, lr=0.01, epochs=20):
    print("[train_neural_network] Training Neural Network (SGD)...")
    nn = NeuralNetworkSGD(X_train.shape[1], hidden_size, lr, epochs)
    nn.fit(X_train, y_train)
    print()
    return nn


# ============================================================
# SECTION 5-C: XGBOOST BASELINE
# ============================================================

def train_xgboost(X_train, y_train):
    print("[train_xgboost] Training XGBoost...")
    m = XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.1,
                      eval_metric='logloss', use_label_encoder=False,
                      random_state=42, verbosity=0)
    m.fit(X_train, y_train)
    print("[train_xgboost] Done.\n")
    return m


# ============================================================
# SECTION 6: EVALUATION
# ============================================================

def evaluate_models(models: dict, X_test, y_test) -> pd.DataFrame:
    rows = []
    print("=" * 65)
    print("  MODEL EVALUATION — DETAILED REPORTS")
    print("=" * 65)

    for name, model in models.items():
        y_pred = model.predict(X_test)
        y_prob = model.predict_proba(X_test)
        if hasattr(y_prob, 'ndim') and y_prob.ndim == 2:
            y_prob = y_prob[:, 1]

        auc_sc = roc_auc_score(y_test, y_prob)
        rows.append({
            'Model':     name,
            'Accuracy':  round(accuracy_score(y_test, y_pred), 4),
            'Precision': round(precision_score(y_test, y_pred, zero_division=0), 4),
            'Recall':    round(recall_score(y_test, y_pred, zero_division=0), 4),
            'F1-Score':  round(f1_score(y_test, y_pred, zero_division=0), 4),
            'ROC-AUC':   round(auc_sc, 4),
        })
        print(f"\n── {name} ──────────────────────────────")
        print(classification_report(y_test, y_pred,
              target_names=['Legit', 'Fraud'], zero_division=0))
        print(f"  ROC-AUC : {auc_sc:.4f}")

    summary = pd.DataFrame(rows).set_index('Model')

    print("\n" + "=" * 65)
    print("  COMPARISON TABLE")
    print("=" * 65)
    display(summary.style
            .background_gradient(cmap='RdYlGn', subset=['ROC-AUC', 'Recall'])
            .format('{:.4f}'))

    best = summary['ROC-AUC'].idxmax()
    print(f"\n🏆 Best by ROC-AUC → {best}  ({summary.loc[best,'ROC-AUC']:.4f})")
    print("   (ROC-AUC is preferred over accuracy for imbalanced fraud data)\n")
    return summary


# ============================================================
# SECTION 7: ROC CURVES — all models on one plot
# ============================================================

def plot_roc_curves(models: dict, X_test, y_test):
    plt.figure(figsize=(8, 6))
    palette = ['royalblue', 'darkorange', 'forestgreen']

    for (name, model), color in zip(models.items(), palette):
        y_prob = model.predict_proba(X_test)
        if hasattr(y_prob, 'ndim') and y_prob.ndim == 2:
            y_prob = y_prob[:, 1]
        fpr, tpr, _ = roc_curve(y_test, y_prob)
        plt.plot(fpr, tpr, color=color, lw=2.2,
                 label=f"{name}  (AUC={auc(fpr,tpr):.3f})")

    plt.plot([0,1],[0,1], 'k--', lw=1, label='Random (AUC=0.5)')
    plt.fill_between([0,1],[0,1], alpha=0.05, color='grey')
    plt.xlabel("False Positive Rate", fontsize=12)
    plt.ylabel("True Positive Rate", fontsize=12)
    plt.title("ROC Curves — Credit Card Fraud Detection", fontsize=14, fontweight='bold')
    plt.legend(loc='lower right', fontsize=10)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    _save_and_show('3_roc_curves.png')


# ============================================================
# SECTION 8: TRAINING LOSS CURVES
# ============================================================

def plot_training_losses(lr_model, nn_model):
    fig, axes = plt.subplots(1, 2, figsize=(13, 4))
    fig.suptitle("Training Loss Curves", fontsize=14, fontweight='bold')

    axes[0].plot(lr_model.losses, color='darkorange', lw=1.8)
    axes[0].set_title("Logistic Regression — Full-Batch GD", fontsize=12)
    axes[0].set_xlabel("Iteration");  axes[0].set_ylabel("BCE Loss")
    axes[0].grid(alpha=0.3)
    axes[0].annotate("Smooth descent\n(full-batch GD)",
                     xy=(len(lr_model.losses)//2,
                         lr_model.losses[len(lr_model.losses)//2]),
                     xytext=(len(lr_model.losses)*0.6, lr_model.losses[0]*0.85),
                     arrowprops=dict(arrowstyle='->', color='black'),
                     fontsize=9)

    axes[1].plot(nn_model.losses, color='forestgreen', lw=1.8)
    axes[1].set_title("Neural Network — SGD (per-sample updates)", fontsize=12)
    axes[1].set_xlabel("Epoch");  axes[1].set_ylabel("BCE Loss")
    axes[1].grid(alpha=0.3)

    plt.tight_layout()
    _save_and_show('4_training_losses.png')


# ============================================================
# SECTION 9: TOEPLITZ MATRIX VISUALISATION  (Unit 1 concept)
# ============================================================

def plot_toeplitz_concept(feature_series: np.ndarray, lags: int = 3):
    """Show a small snippet of the Toeplitz-structured feature matrix."""
    snippet = feature_series[:8]
    rows = []
    for t in range(lags, len(snippet)):
        rows.append([snippet[t - k] for k in range(lags + 1)])

    T = np.array(rows, dtype=float)
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    fig.suptitle("Toeplitz / Sliding-Window Feature Matrix (Unit 1)",
                 fontsize=13, fontweight='bold')

    # Left: heatmap of the matrix
    sns.heatmap(T, annot=True, fmt='.1f', cmap='Blues',
                ax=axes[0], cbar=False,
                xticklabels=[f't-{k}' if k else 't' for k in range(lags+1)],
                yticklabels=[f'row {i}' for i in range(T.shape[0])])
    axes[0].set_title("Rows = [x(t), x(t-1), x(t-2), x(t-3)]")
    axes[0].set_xlabel("Lag column")

    # Right: annotated signal with arrows showing the window
    axes[1].plot(snippet, 'o-', color='steelblue', lw=2, ms=8)
    for k in range(1, lags+1):
        axes[1].annotate('', xy=(lags, snippet[lags]),
                         xytext=(lags-k, snippet[lags-k]),
                         arrowprops=dict(arrowstyle='->', color='crimson',
                                         lw=1.3, linestyle='dashed'))
    axes[1].set_title("Shift-invariant sliding window on signal")
    axes[1].set_xlabel("Time index t");  axes[1].set_ylabel("Value")
    axes[1].grid(alpha=0.3)

    plt.tight_layout()
    _save_and_show('5_toeplitz_concept.png')


# ============================================================
# MAIN PIPELINE
# ============================================================

def main():
    print("\n" + "═"*60)
    print("  CREDIT CARD FRAUD DETECTION — ML PIPELINE")
    print("═"*60 + "\n")

    # 1. Load
    df = load_data('/content/creditcard.csv')

    # 2. EDA
    selected_features = eda(df)
    df_sel = df[selected_features + ['Class']].copy()

    # 3. Toeplitz lag features
    toeplitz_cols = [c for c in ['Amount', 'V17', 'V14', 'V12']
                     if c in df_sel.columns][:2]
    df_feat = create_toeplitz_features(df_sel, columns=toeplitz_cols, lags=3)

    # Visualise the Toeplitz concept using the first toeplitz column
    plot_toeplitz_concept(df_feat[toeplitz_cols[0]].values[:20], lags=3)

    feature_cols = [c for c in df_feat.columns if c != 'Class']
    print(f"[main] Total features after Toeplitz engineering: {len(feature_cols)}\n")

    # 4. Preprocess
    X_train, X_test, y_train, y_test, scaler = preprocess_data(
        df_feat, feature_cols)

    # 5. Train
    lr_model  = train_logistic_gd(X_train, y_train, lr=0.1, n_iter=800)
    nn_model  = train_neural_network(X_train, y_train,
                                      hidden_size=16, lr=0.01, epochs=20)
    xgb_model = train_xgboost(X_train, y_train)

    # 6. Evaluate
    models = {
        'Logistic Regression (GD)': lr_model,
        'Neural Network (SGD)':     nn_model,
        'XGBoost':                  xgb_model,
    }
    summary = evaluate_models(models, X_test, y_test)

    # 7. Plots
    plot_roc_curves(models, X_test, y_test)
    plot_training_losses(lr_model, nn_model)

    # 8. Summary
    print("\n" + "═"*60)
    print(f"  ✅  All outputs saved to: {OUTPUT_DIR}/")
    print("  Files produced:")
    for f in sorted(os.listdir(OUTPUT_DIR)):
        print(f"    • {f}")
    print("═"*60)

    return lr_model, nn_model, xgb_model, scaler, feature_cols

global lr_model, nn_model, xgb_model, scaler, feature_cols
lr_model, nn_model, xgb_model, scaler, feature_cols = main()
```
